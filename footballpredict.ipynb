```python
import random

# Team data (players and strength)
teams = {
    "Real Madrid": {"players": ["Ronaldo", "Benzema", "Modric"], "strength": 90},
    "Barcelona": {"players": ["Messi", "Pedri", "Lewandowski"], "strength": 88},
    "Atletico Madrid": {"players": ["Griezmann", "Morata", "Koke"], "strength": 85},
    "Sevilla": {"players": ["Rakitic", "Ocampos"], "strength": 80},
    "Valencia": {"players": ["Guedes", "Duro"], "strength": 78}
}

# Function to predict the match
def predict_match(team1, team2):
    if team1 not in teams or team2 not in teams:
        print("Invalid team names.")
        return

    # Goal prediction based on team strength
    score1, score2 = random.randint(0, teams[team1]["strength"] // 20), random.randint(0, teams[team2]["strength"] // 20)

    # Randomly choose players to score
    scorers1, scorers2 = random.choices(teams[team1]["players"], k=score1), random.choices(teams[team2]["players"], k=score2)

    # Display result
    print(f"\n{team1} {score1} - {score2} {team2}")
    print(f"{team1} goals by: {', '.join(scorers1) if scorers1 else 'No one'}")
    print(f"{team2} goals by: {', '.join(scorers2) if scorers2 else 'No one'}")
    print(f"{team1} wins!" if score1 > score2 else f"{team2} wins!" if score2 > score1 else "It's a draw!")

# Get team names from the user
team1 = input("Enter first team: ")
team2 = input("Enter second team: ")

# Call the prediction function
predict_match(team1, team2)
```
